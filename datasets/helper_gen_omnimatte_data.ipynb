```python
# movie
# mask * rgb_removed + (1 - mask) * rgb_full is our final input for training (named as rgb_full)
# orig_dir = '/playpen-nas-ssd/luchao/projects/gen-omnimatte-public/datasets/train-casper/kubric_orig'
# target_dir = '/playpen-nas-ssd/luchao/projects/gen-omnimatte-public/datasets/train-casper/kubric'
orig_dir = '/playpen-nas-ssd/luchao/data/OmnimatteRF/movie'
target_dir = '/playpen-nas-ssd/luchao/projects/gen-omnimatte-public/datasets/train-casper/movie'
os.makedirs(target_dir, exist_ok=True)

for subdir in sorted(os.listdir(orig_dir)):
    if subdir in ['dodge']:
        # not much effects in these data
        continue
    if not os.path.isdir(os.path.join(orig_dir, subdir)):
        continue
    
    os.makedirs(os.path.join(target_dir, subdir, 'trimask'), exist_ok=True)
    os.makedirs(os.path.join(target_dir, subdir, 'input'), exist_ok=True)
    os.makedirs(os.path.join(target_dir, subdir, 'bg'), exist_ok=True)

    mask_parent_dir = os.path.join(orig_dir, subdir, 'masks', 'mask')  # may contain 00/01/02/...
    if not os.path.isdir(mask_parent_dir):
        continue
    # Collect and binarize all sub-mask sequences
    mask_sequences = []
    max_num_frames = 0
    target_height, target_width = None, None
    for mask_idx in sorted(os.listdir(mask_parent_dir)):
        current_mask_dir = os.path.join(mask_parent_dir, mask_idx)
        if not os.path.isdir(current_mask_dir):
            continue
        sequence = _read_video_from_dir(current_mask_dir)  # (T,H,W) or (T,H,W,3)
        if sequence.ndim == 4:  # RGB -> single channel (take max to be inclusive)
            sequence = sequence.max(axis=-1)
        # Binarize; mask assets are typically 0/255
        sequence = (sequence > 127).astype(np.uint8)
        if target_height is None or target_width is None:
            target_height, target_width = sequence.shape[1], sequence.shape[2]
        elif sequence.shape[1] != target_height or sequence.shape[2] != target_width:
            # Resize spatially with nearest neighbor to preserve hard edges
            sequence = np.stack(
                [cv2.resize(frame, (target_width, target_height), interpolation=cv2.INTER_NEAREST) for frame in sequence],
                axis=0,
            )
        mask_sequences.append(sequence)
        max_num_frames = max(max_num_frames, sequence.shape[0])
    if not mask_sequences:
        continue
    # Merge by logical OR across all sub-masks, padding shorter sequences with zeros
    merged_mask = np.zeros((max_num_frames, target_height, target_width), dtype=np.uint8)
    for sequence in mask_sequences:
        num_frames = sequence.shape[0]
        merged_mask[:num_frames] = np.maximum(merged_mask[:num_frames], sequence)

    file_paths = sorted(list(glob.glob(os.path.join(orig_dir, subdir, 'bg_rgb_1x', '*.png'))))
    file_names = [os.path.basename(file_path) for file_path in file_paths] # the number may not be continuous

    # Repeat mask 3 times along new last axis to get shape (frames, height, width, channels=3)
    mask = np.stack([merged_mask] * 3, axis=-1)
    bg = _read_video_from_dir(os.path.join(orig_dir, subdir, 'bg_rgb_1x'))
    input_img = _read_video_from_dir(os.path.join(orig_dir, subdir, 'rgb_1x'))
    
    has_alpha = False
    if input_img.shape[-1] == 4 or bg.shape[-1] == 4:
        has_alpha = True
        input_img = input_img[..., :3]
        bg = bg[..., :3]
    output_image = mask * input_img + (1 - mask) * bg

    assert len(file_names) == output_image.shape[0]
    for i in range(output_image.shape[0]):
        media.write_image(os.path.join(target_dir, subdir, 'input', file_names[i]), output_image[i]) # slap comp
        # media.write_image(os.path.join(target_dir, subdir, 'trimask', file_names[i]), merged_mask[i] * 255) # binary mask
        if has_alpha:
            # per frame saving
            media.write_image(os.path.join(target_dir, subdir, 'bg', file_names[i]), input_img[i])
        else:
            # directly copy raw
            shutil.copy(os.path.join(orig_dir, subdir, 'rgb_1x', file_names[i]), os.path.join(target_dir, subdir, 'bg', file_names[i]))
    # break
```

```python
# omnimatte_zero
# mask can be get using grounding dino script in /playpen-nas-ssd/luchao/projects/Grounded-SAM-2/helper_gdino_sam.ipynb
# mask can also be get using sam2 notebook
orig_dir = '/playpen-nas-ssd/luchao/data/omnimatte_zero'
target_dir = '/playpen-nas-ssd/luchao/projects/gen-omnimatte-public/datasets/train-casper/omnimatte_zero_ours'

for subdir in sorted(os.listdir(orig_dir)):
    # original (effect) is video.mp4
    # edited (no effect) is omnimatteZero.mp4 or omnimatteZero.mov
    if os.path.isdir(os.path.join(orig_dir, subdir)):
        os.makedirs(os.path.join(target_dir, subdir), exist_ok=True)
        effect_video_path = os.path.join(orig_dir, subdir, 'video.mp4')
        no_effect_video_path = os.path.join(orig_dir, subdir, 'omnimatteZero.mp4') if os.path.exists(os.path.join(orig_dir, subdir, 'omnimatteZero.mp4')) else os.path.join(orig_dir, subdir, 'omnimatteZero.mov')
        if os.path.exists(effect_video_path) and os.path.exists(no_effect_video_path):
            # shutil.copy(effect_video_path, os.path.join(target_dir, subdir, 'rgb_removed.mp4'))
            # shutil.copy(no_effect_video_path, os.path.join(target_dir, subdir, 'rgb_full.mp4'))
            effect_video = media.read_video(effect_video_path)
            no_effect_video = media.read_video(no_effect_video_path)
            fps_effect = effect_video.metadata.fps
            fps_no_effect = no_effect_video.metadata.fps
            if fps_effect != fps_no_effect:
                print(f'WARNING:{subdir} has different fps, {fps_effect} vs {fps_no_effect}')
                # use either fps 15 or 30 depending whichever is closer
                if abs(fps_effect - 15) < abs(fps_effect - 30):
                    fps = 15
                else:
                    fps = 30
            effect_shape = effect_video.shape # [t, h, w, 3]
            no_effect_shape = no_effect_video.shape
            if effect_shape[0] != no_effect_shape[0]:
                print(f'{subdir} has {effect_shape[0]} frames, {no_effect_shape[0]} frames')
                min_len = min(effect_shape[0], no_effect_shape[0])
                effect_video = effect_video[:min_len]
                no_effect_video = no_effect_video[:min_len]
            # usually they do not have the same (hw)
            # we resize them to the same shape (larger one)
            effect_h, effect_w = effect_shape[1], effect_shape[2]
            no_effect_h, no_effect_w = no_effect_shape[1], no_effect_shape[2]
            h, w = max(effect_h, no_effect_h), max(effect_w, no_effect_w)
            if h == effect_h and w == effect_w:
                # effect is larger, change no_effect_video
                no_effect_video = media.resize_video(no_effect_video, (h, w))
                media.write_video(os.path.join(target_dir, subdir, 'rgb_full.mp4'), no_effect_video, fps=fps)
                media.write_video(os.path.join(target_dir, subdir, 'rgb_removed.mp4'), effect_video, fps=fps)
            else:
                effect_video = media.resize_video(effect_video, (h, w))
                media.write_video(os.path.join(target_dir, subdir, 'rgb_removed.mp4'), effect_video, fps=fps)
                media.write_video(os.path.join(target_dir, subdir, 'rgb_full.mp4'), no_effect_video, fps=fps)

            # save images for sam2
            sam2_dir = os.path.join(target_dir, subdir, 'sam2')
            os.makedirs(sam2_dir, exist_ok=True)
            for i in range(effect_video.shape[0]):
                img = effect_video[i]
                media.write_image(os.path.join(sam2_dir, f'{i}.jpg'), img)
        else:
            print(f'{subdir} does not have both effect and no effect videos')
            continue
```

```python
# rose
parent_dir = '/playpen-nas-ssd/luchao/data/ROSE-Dataset'

orig_dir_synthetic_videos = '/playpen-nas-ssd/luchao/data/HYouTube/synthetic_composite_videos'
mask_dir = '/playpen-nas-ssd/luchao/data/HYouTube/foreground_mask'

target_dir = f'/playpen-nas-ssd/luchao/projects/gen-omnimatte-public/datasets/train-casper/rose_ours'


def helper(env_name, original_video_path, edited_video_path, mask_path):
    target_name = env_name + '_' + os.path.splitext(os.path.basename(original_video_path))[0]
    if os.path.exists(os.path.join(target_dir, target_name, 'rgb_full.mp4')):
        return
    os.makedirs(os.path.join(target_dir, target_name), exist_ok=True)

    # copy the edited video
    # shutil.copy(edited_video_path, os.path.join(target_dir, target_name, 'input_video.mp4'))
    # shutil.copy(mask_path, os.path.join(target_dir, target_name, 'trimask_00.mp4'))
    shutil.copy(original_video_path, os.path.join(target_dir, target_name, 'rgb_removed.mp4'))
    # shutil.copy(mask_path, os.path.join(target_dir, target_name, 'mask.mp4'))

    # rgb_full is mask * unedited + (1 - mask) * edited
    mask = media.read_video(mask_path)
    fps = mask.metadata.fps
    threshold = 127
    mask = np.where(mask > threshold, 1, 0).astype(np.uint8)
    # Save the processed mask as 'mask.mp4' in the target directory
    media.write_video(os.path.join(target_dir, target_name, 'mask.mp4'), mask * 255, fps=fps)
    edited = media.read_video(edited_video_path)
    unedited = media.read_video(original_video_path)
    rgb_full = mask * unedited + (1 - mask) * edited
    media.write_video(os.path.join(target_dir, target_name, 'rgb_full.mp4'), rgb_full, fps=fps)
    del rgb_full, mask, edited, unedited

    # print(f'processed {target_name}')
    return

os.makedirs(target_dir, exist_ok=True)
tasks = []
for env_name in sorted(os.listdir(parent_dir)):
    if env_name == '.git':
        continue
    if not os.path.isdir(os.path.join(parent_dir, env_name)):
        continue
    edited_dir = os.path.join(parent_dir, env_name, env_name + '_Edited')
    mask_dir = os.path.join(parent_dir, env_name, env_name + '_Masked')
    original_dir = os.path.join(parent_dir, env_name, env_name + '_Unedited')
    for video_name in os.listdir(edited_dir):
        original_video_path = os.path.join(original_dir, video_name)
        edited_video_path = os.path.join(edited_dir, video_name)
        mask_path = os.path.join(mask_dir, video_name)
        if os.path.exists(original_video_path) and os.path.exists(edited_video_path) and os.path.exists(mask_path):
            tasks.append((env_name, original_video_path, edited_video_path, mask_path))

with Pool(processes=mp.cpu_count() // 2) as pool:
    pool.starmap(helper, tasks)
    # r = list(tqdm.tqdm(pool.starmap(helper, tasks), total=len(tasks), dynamic_ncols=True))
```

```python
# hyoutube
from PIL import Image
orig_dir_real_videos = '/playpen-nas-ssd/luchao/data/HYouTube/real_videos'
orig_dir_synthetic_videos = '/playpen-nas-ssd/luchao/data/HYouTube/synthetic_composite_videos'
mask_dir = '/playpen-nas-ssd/luchao/data/HYouTube/foreground_mask'
object_index = 'object_5'
target_dir = f'/playpen-nas-ssd/luchao/projects/gen-omnimatte-public/datasets/train-casper/hyoutube-{object_index}'


def helper(subdir):
    os.makedirs(os.path.join(target_dir, subdir), exist_ok=True)
    os.makedirs(os.path.join(target_dir, subdir, 'input'), exist_ok=True)
    os.makedirs(os.path.join(target_dir, subdir, 'trimask'), exist_ok=True)
    os.makedirs(os.path.join(target_dir, subdir, 'bg'), exist_ok=True)

    for file in os.listdir(os.path.join(orig_dir_synthetic_videos, subdir, object_index)):
        file_name = os.path.splitext(file)[0]
        # copy the synthetic video
        img = Image.open(os.path.join(orig_dir_synthetic_videos, subdir, object_index, file_name + '.jpg'))
        img.save(os.path.join(target_dir, subdir, 'input', file_name + '.png'), 'PNG')
        # copy the original real video
        img = Image.open(os.path.join(orig_dir_real_videos, subdir, file_name + '.jpg'))
        img.save(os.path.join(target_dir, subdir, 'bg', file_name + '.png'), 'PNG')
        # copy the mask
        shutil.copy(os.path.join(mask_dir, subdir, object_index, file_name + '.png'), os.path.join(target_dir, subdir, 'trimask', file_name + '.png'))
    print(f'processed {subdir}')
    return

os.makedirs(target_dir, exist_ok=True)
subdirs = []
for subdir in sorted(os.listdir(orig_dir_real_videos)):
    if not os.path.isdir(os.path.join(orig_dir_real_videos, subdir)):
        continue
    # check if subdir is in orig_dir_synthetic_videos
    # * we use "object_0" for now
    if not os.path.isdir(os.path.join(orig_dir_synthetic_videos, subdir, object_index)):
        continue
    if not os.path.isdir(os.path.join(mask_dir, subdir, object_index)):
        continue
    subdirs.append(subdir)

with Pool(processes=mp.cpu_count() - 2) as pool:
    pool.map(helper, subdirs)
```
